```python
#enter 5 number and find greatest
no=5
greatest=0
while True:
    for i in range(no):
        n1=int(input(f"Enter the num {i+1}:")) #68
        if n1>greatest:
            greatest=n1 #gt=68
    print("greatest number is:",greatest)
    break
            
        
```

```python
# greatestnum[34 42 24]
num=[]
for i in range(4):
    n1=int(input(f"Enter the number {i+1}"))
    num.append(n1)  #[32 ]
greatest=max(num)
print("Greatest number:",greatest)
total=sum(num)
length=len(num)
print("average:",total/length)
    
```

```python
#to reverse a string
reverse=""
string=input("Enter a string:") #kavya
for i in string:
    reverse=i+reverse
print("reverse string is:",reverse)
count=0
for s in string.lower():
    if s in "aeiou":
        count +=1
print("vowels count:",count)
```

```python
#duplicate values in the list
l1=[]
for i in range(5):
    num=input(f"Enter the numbers for the list:{i+1}")
    l1.append(num)
l2=[]
print("List=",l1)   #[89 78 56 78 66]
for i in l1:
    if l1.count(i)>1 and (i not in l2) :
        l2.append(i)
print("Duplicate values:",l2)
```

```python
#factorial of a number
fact=1
n=int(input("Enter a number:"))
for i in range(1,n+1):
    fact*=i
print(fact)
    
    
```

```python
#find frequency of numbers
l1=[]
for i in range(5):
    num=int(input("Enter the numbers:"))
    l1.append(num)
l2={}
for i in l1:
    l2[i]=l1.count(i)
print(l2)
    
```
