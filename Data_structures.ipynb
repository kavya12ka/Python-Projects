```python
s1={3,6,9,10,24,50,60,70,100}
s2={6,20,30,40,50,60,74,120}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s2.difference(s1))
print(s2.symmetric_difference(s1))
s1.symmetric_difference_update(s2)
print(s1)
```
