```python
## Program for a number guessing game
import random
num=random.randint(1,100)
attempt=0
max_life=2
life_lost=0
while True:
    n1=int(input("Enter the number"))
    if n1>100 or n1<0:
        if life_lost==max_life-1:
            print("All lifes lost")
            break
        else:
            print("One life left")
            life_lost+=1
            continue
    attempt+=1
    if num==n1:
        print("Congragulation!!you won the game in ",attempt,"attempts" )
        break
    else:
        print("Sorry!! Try Again")
        if n1>num:
            print("The is smaller")
        else:
            print("The is greater")
                
```
