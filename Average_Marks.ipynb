```python
#find the Average marks of students 
check=input("If you want to calculate avg(y to continue):")
sub=int(input("enter the number of subjects:"))
while check=="y" or check=="Y":
    total_marks=0
    for i in range(sub):
        marks=float(input(f"enter the marks in subject {i+1}:"))
        total_marks +=marks
    print("avg=",total_marks/sub)
    print("total marks",total_marks)
    check1=input("If you want to calculate avg(y to continue):")
    if check1!="y":
        print("end")
        break
    else:
        continue  
```
